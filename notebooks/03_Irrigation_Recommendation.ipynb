```python
def irrigation_advice(temp, humidity, rainfall):
    if rainfall > 20:
        return "No irrigation needed today."
    elif temp > 35 and humidity < 40:
        return "Use drip irrigation and water twice today."
    elif temp > 30:
        return "Use sprinkler irrigation in the evening."
    elif humidity > 80:
        return "Water lightly once in the morning."
    else:
        return "Water once a day using drip irrigation."
```

```python
temperature = float(input("Temperature (°C): "))
humidity = float(input("Humidity (%): "))
rainfall = float(input("Rainfall (mm): "))
result = irrigation_advice(temperature, humidity, rainfall)
print(result)
```
